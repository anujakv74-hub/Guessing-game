Guessing game

```
import random

# Number Guessing Game

print("===================================")
print("      Welcome to Guess the Number!")
print("===================================")

# Game settings
lower_limit = 1
upper_limit = 100
max_attempts = 7

# Generate a random number
secret_number = random.randint(lower_limit, upper_limit)

print(f"\nI have chosen a number between {lower_limit} and {upper_limit}.")
print(f"You have {max_attempts} attempts to guess it.")

# Game loop
for attempt in range(1, max_attempts + 1):
    print(f"\nAttempt {attempt} of {max_attempts}")

    try:
        guess = int(input("Enter your guess: "))

        if guess < lower_limit or guess > upper_limit:
            print(f"Please enter a number between {lower_limit} and {upper_limit}.")
            continue

        if guess < secret_number:
            print("Too low! Try again.")
        elif guess > secret_number:
            print("Too high! Try again.")
        else:
            print(f"\nCongratulations! You guessed the number {secret_number} correctly!")
            print(f"You won in {attempt} attempt(s).")
            break

    except ValueError:
        print("Invalid input! Please enter a whole number.")

else:
    print("\nGame Over!")
    print(f"You've used all {max_attempts} attempts.")
    print(f"The correct number was {secret_number}. Better luck next time!")

print("\nThank you for playing!")
```
